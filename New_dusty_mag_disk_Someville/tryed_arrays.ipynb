```python
P0=825 # mbar
T0=30+273.15 # K
e_tau=6.11 # mbar
T_tau=273.15 # K
L_v=2.5*10**6 # J/kg
R_v=461 # J/kg/K

def saturation_vapor_pressure(T):
    return e_tau * np.exp(L_v/R_v * (1/T_tau - 1/T))

def addiabatic_pressure(T):
    return P0 * (T/T0)**(3.5)*(0.0385) # obtained with the mixing ratio of 25g/kg

# we want to find the temperature at which the saturation vapor pressure equals the adiabatic pressure
from scipy.optimize import fsolve
def equation(T):
    return saturation_vapor_pressure(T) - addiabatic_pressure(T)
T_solution = fsolve(equation, 300) # Initial guess of 300 K
print('Temperature at which saturation vapor pressure equals adiabatic pressure:', T_solution[0]-273.15, '°C')
print('Saturation vapor pressure at this temperature:', saturation_vapor_pressure(T_solution[0]), 'mbar')
print('Adiabatic pressure at this temperature:', addiabatic_pressure(T_solution[0]), 'mbar')
```

```python
L=2.5*10**6 # J/kg
Rd=287 # J/kg/K
Rv=461 # J/kg/K
epsilon=Rd/Rv
cp=1004 # J/kg/K
T_cloud=23.5+273.15 # K
r_c=0.025 # saturation mixing ratio 
Gamma_dry=9.8 # K/km

semiadiabatic_lapse_rate=(1+r_c*L/(Rd*T_cloud))/(1+r_c*L**2*epsilon/(Rv*cp*T_cloud**2))*Gamma_dry

print('Semi-adiabatic lapse rate:', semiadiabatic_lapse_rate, 'K/km')
```
